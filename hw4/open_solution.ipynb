```python
import pandas as pd
import numpy as np
from collections import defaultdict

obj = pd.read_pickle(r'solutions.dat')
solution = list(obj.keys())[13] # get index of solution
solution
```

```python
assign = np.where(obj[solution]==1)
ta = defaultdict(list)
sect = defaultdict(list)
for i in range(len(assign[0])):
    ta[int(assign[0][i])].append(int(assign[1][i]))
    sect[int(assign[1][i])].append(int(assign[0][i]))
    
print(f'TAs are assigned to these sections:\n{ta}\nSections have these TAs assigned to them:\n{sect}')
```
